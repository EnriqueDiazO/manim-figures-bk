# Why is $C$ only real-linear?

Here $C$ means complex conjugation: $(Cf)(t)=\overline{f(t)}$. The animation reflects a complex-valued curve across the real axis and then checks the scalar rule.

```python
# ============================================================
# USER PARAMETERS
# ============================================================
# ANIMATION PARAMETERS
FPS = 24
SPEED_FACTOR = 1.0
BASE_DURATION = 6.0
```

## Mathematical checkpoint

Write $f(t)=u(t)+iv(t)$ with real $u,v$. Then $Cf=u-iv$. For a scalar $\lambda$, $C(\lambda f)=\overline{\lambda}\,Cf$; this equals $\lambda Cf$ for real $\lambda$ but not generally for complex $\lambda$. The symbol $C$ in this notebook is conjugation; in the branch-interaction grid, C labels a cusp branch.

```python
from pathlib import Path
import sys
project_dir = Path.cwd()
if project_dir.name == "notebooks":
    project_dir = project_dir.parent
elif not (project_dir / "src" / "geometry.py").exists():
    project_dir = project_dir / "karlovich_2025_cusps"
if not (project_dir / "src" / "geometry.py").exists():
    raise FileNotFoundError("Run this notebook from its project folder or the repository root.")
sys.path.insert(0, str(project_dir / "src"))
output_dir = project_dir / "outputs"

import numpy as np
from render_complex_conjugation import render_complex_conjugation
assert FPS > 0 and SPEED_FACTOR > 0 and BASE_DURATION > 0
sample_f = 1 + 2j
lam = 1j
assert np.conj(lam * sample_f) == np.conj(lam) * np.conj(sample_f)
assert np.conj(lam * sample_f) != lam * np.conj(sample_f)
print("C(i f) =", np.conj(lam * sample_f), "; i C(f) =", lam * np.conj(sample_f))
```

## Watch the reflection

The dashed curve is $f(t)$; the moving curve changes its imaginary coordinate from $v(t)$ to $-v(t)$ while $u(t)$ stays fixed. The scalar identity and the real/complex linearity verdict appear only after the reflection.

```python
from IPython.display import Image, display
assets = render_complex_conjugation(output_dir, fps=FPS,
                                    duration=BASE_DURATION, speed_factor=SPEED_FACTOR)
for key in ("gif", "mp4", "png", "pdf"):
    print(f"{key}: {assets[key].relative_to(project_dir)}")
print(f"{assets['frames']} frames; {assets['duration_seconds']:.2f} seconds")
display(Image(filename=str(assets["png"])))
```

## Read the picture

Reflection changes the sign of the imaginary part. The identity $C(\lambda f)=\overline\lambda Cf$ is the reason a problem involving $C$ must be represented over real scalars before it can be doubled into a complex-linear matrix problem.
