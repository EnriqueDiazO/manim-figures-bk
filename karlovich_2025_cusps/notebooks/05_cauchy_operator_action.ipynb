```python
# USER PARAMETERS
X_MIN = -3.5                 # left edge of both fixed graph axes
X_MAX = 3.5                  # right edge of both fixed graph axes
N_PLOT = 701                 # samples in the displayed input and output
NUMERICAL_TOLERANCE = 2e-5  # independent PV check must beat this error
FPS = 24                     # video frames per second
SPEED_FACTOR = 1.0           # larger values shorten the animation
HOLD_INPUT = 1.0 / SPEED_FACTOR
REVEAL_OUTPUT = 2.8 / SPEED_FACTOR
HOLD_COMPARISON = 2.0 / SPEED_FACTOR
HOLD_END = 0.8 / SPEED_FACTOR
OUTPUT_NAME = "cauchy_operator_action"
OUTPUT_GIF = OUTPUT_NAME + ".gif"
OUTPUT_MP4 = OUTPUT_NAME + ".mp4"
OUTPUT_PNG = OUTPUT_NAME + ".png"
OUTPUT_PDF = OUTPUT_NAME + ".pdf"
```

## Mathematical convention and independent check

Here
\[
Hf(x)=\frac{1}{\pi}\operatorname{p.v.}\int_{\mathbb R}\frac{f(\tau)}{x-\tau}\,d\tau,
\qquad f(x)=e^{-x^2},
\qquad Hf(x)=\frac{2}{\sqrt{\pi}}D(x),
\]
where \(D(x)=e^{-x^2}\int_0^x e^{u^2}\,du\). For the orientation and Cauchy normalization of Karlovich (2025), (3.1), \(Sf=iHf\) on this real-line model. The plotted output is the real-valued \(Hf\).

The source code calculates \(D\) by ordinary quadrature and checks it independently by removing the principal-value singularity:
\[
\operatorname{p.v.}\int_{-L}^{L}\frac{f(\tau)}{x-\tau}\,d\tau
=\int_{-L}^{L}\frac{f(\tau)-f(x)}{x-\tau}\,d\tau
+f(x)\log\frac{x+L}{L-x}.
\]
For the Gaussian, the omitted tails at \(L=9\) are negligible at the seven check points. This verification checks both the sign and the factor \(1/\pi\).

```python
from pathlib import Path
import sys
from IPython.display import display
import matplotlib.pyplot as plt

project_dir = Path.cwd()
if project_dir.name == "notebooks":
    project_dir = project_dir.parent
elif not (project_dir / "src" / "render_cauchy_operator_action.py").exists():
    project_dir = project_dir / "karlovich_2025_cusps"
if not (project_dir / "src" / "render_cauchy_operator_action.py").exists():
    raise FileNotFoundError("Open the notebook from its project or repository folder.")
sys.path.insert(0, str(project_dir / "src"))

from render_cauchy_operator_action import (
    ActionConfig, render_assets, render_frame, verify_normalization
)

config = ActionConfig(
    x_min=X_MIN, x_max=X_MAX, n_plot=N_PLOT, fps=FPS,
    hold_input=HOLD_INPUT, reveal_output=REVEAL_OUTPUT,
    hold_comparison=HOLD_COMPARISON, hold_end=HOLD_END,
)
check = verify_normalization(NUMERICAL_TOLERANCE)
print("Python:", sys.executable, sys.version.split()[0])
print("Independent principal-value check:", check)
assert check["max_abs_error"] < NUMERICAL_TOLERANCE
```

## How to read the result

The choice of \(H\) fixes a sign convention. Under the article's real-line Cauchy convention, the complex output is \(Sf=iHf\); the violet curve is therefore its imaginary part. The numerical check above is part of this illustration only. For the curve \(\Gamma\), the article proceeds through the exact transported kernels in (3.6), compact models, and local symbols. The present example stops before that reduction.
